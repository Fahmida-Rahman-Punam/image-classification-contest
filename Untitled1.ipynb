```python
import os
train_files= os.listdir('/content/drive/MyDrive/train')
print(len(train_files))
print(train_files[:5])
```

```python
import tensorflow as tf
train_data=tf.keras.utils.image_dataset_from_directory('/content/drive/MyDrive/train',image_size=(128,128),batch_size=32)
```

```python
base_model=tf.keras.applications.MobileNetV2(input_shape=(128,128,3),include_top=False,weights='imagenet')
base_model.trainable = False
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(10,activation='softmax')
])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.fit(train_data,epochs=1)
```

```python
import os
import glob
import numpy as np
import pandas as pd
import tensorflow as tf

test_dir = "/content/drive/MyDrive/test"

image_paths = glob.glob(os.path.join(test_dir,"**","*.jpg"),recursive=True)+\
              glob.glob(os.path.join(test_dir,"**","*.png"),recursive=True)+\
              glob.glob(os.path.join(test_dir,"**","*.jpeg"),recursive=True)
print(f"total test images : {len(image_paths)}")

results=[]

for p in sorted(image_paths):
  filename=os.path.basename(p)
  image_id=os.path.splitext(filename)[0]

  img = tf.keras.utils.load_img(p,target_size=(128,128))
  a = tf.keras.utils.img_to_array(img)
  a = tf.expand_dims(a,0)

  pred = model.predict(a,verbose=0)
  c=np.argmax(pred[0])
  results.append([image_id,c])
sub = pd.DataFrame(results,columns=["id","target"])
try :
      sub['id'] = sub['id'].astype(int)
      sub = sub.sort_values('id')
except ValueError:
      pass
sub.to_csv('/content/submission.csv',index=False)
print(f"done! saved {len(sub)} predictions to /content/submission.csv")
```
